## Find average app Rating in Google Play Store

```python
# Task 9: Find the average app rating in google play store
# First access the Rating column --> Using mean() helps to obtain average rtating of the apps
print('Average App rating =',df['Rating'].mean())
```

## Find the total number of unique category apps

## Find the Category with the Highest Average Rating

```python
# Task 11 : Find the category with the highest average rating
# Group by category and calculate mean ratings sort the values in descending order access the top Category
# This give a series dtype with Category and top rating
# Accessing the index of that series gives us Category name 
rating = df.groupby('Category')['Rating'].mean().sort_values(ascending=False).head(1)
print('Highest average rating of app')
print('---------------------------------------------------------------------------------')
print(rating)
print('---------------------------------------------------------------------------------')
print('Category with highest average rating')
print('---------------------------------------------------------------------------------')
rating.index[0]
```

## Find the number of apps with 5 star rating

```python
# Task 12: Find the number of apps with 5 star rating
# We filter the dataset using condition
# Access the Apps with Rating of 5 retuns a data set --> len() the data gives number of apps
# sum() dataset can be used to find number of apps rated with 5 stars
apps = df[df['Rating']==5]
print('Number of apps achieved 5 star rating in google playstore =',len(apps))
print('------------------------------------------------------------------------')
rating = df['Rating']==5 # Returns bool
print('Number of apps achieved 5 star rating in google playstore =',(rating).sum())
```

## Data Cleaning and Preprocessing

The raw dataset contained inconsistent formats and invalid entries. The following corrections were performed to prepare the data for analysis.

## Find the average value of reviews

```python
# Task 13: Find the average value of reviews
# To obtain average reviews of apps in google playstore we should access the Reviews column
# Replace the strings with numerical values
# Convert the whole Reviews data to Float then find the mean
df['Reviews'].replace('3.0M','3000000',inplace=True)
print('Average value of Reviews =',df['Reviews'].astype(float).mean())
```

## Find the total number of Free paid apps

```python
# Task 14: Find the total number of free paid apps
# To find the number of free apps available in the google playstore
# Access the Type(it describes the app paid/free) column
# Search for free then -->len(free_apps)
free_apps = df[df['Type']=='Free']
print('Number of apps available for free in google playstore =',len(free_apps))
```

```python
print('Total number of paid apps available in google playstore',(df['Type']=='Paid').sum())
```

```python
df['Type'].value_counts()
```

## Find the app with maximum reviews

```python
# Task 15: Find the app with maximum reviews
# Access the Reviews column and then find the max values in reviews column
# Using the max value we can access the App name
# Since it is a series dtype we use values attribute to obtain name
highest = df['Reviews'].max()
print('Maximum reviews =',highest)
print('--------------------------------------------------')
print('App with maximum reviews =', df[df['Reviews']==df['Reviews'].max()]['App'])
```

## Display top 5 apps having highest reviews

```python
# Task 16: Find the top 5 apps having highest reviews
# To obtain the top 5 apps with highest reviews should sort reviews column values in descending order
# Obtain top 5 apps index values using head() function
# Using loc or iloc --> df.iloc[index] gives us apps data
# Then access the app column

ind = df['Reviews'].astype(float).sort_values(ascending=False).head().index
print('Top 5 Apps with highest reviews')
print('-------------------------------------------------------')
print(df.iloc[ind]['App'])
```

## Display top 5 apps having maximum number of installs

```python
# Task 17: Find the top 5 apps with highest number of installs
# Obtain the installs column from dataset
# Convert the Installs column values to int 
# Replace ',' and '+' using replace() function
# Sort the values in installs column, using head() obtain top 5 installs and using index attribute obtain indices of the apps
# Using those indices then obtain the top 5 apps with maximum installs
df['Installs'] = df['Installs'].str.replace(',','')
df['Installs'] = df['Installs'].str.replace('+','')
df['Installs'] = df['Installs'].str.replace('Free','0')
df['Installs'] = df['Installs'].astype(int)
```

```python
index_val = df['Installs'].sort_values(ascending = False).head().index
```

```python
df.loc[index_val]['App']
```

## Find the average rating of free and paid apps

```python
# Task 18: Find the average rating of free and paid apps
# To obtain the average Rating of free and paid apps we use groupby() finction
# groupby('Type') and obtain the mean for Rating column
print('Average rating of Free and paid Apps')
print('-----------------------------------------')
print(df.groupby('Type')['Rating'].mean())
```
